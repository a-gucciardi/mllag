```python
import fitz  # PyMuPDF

# Path to your PDF
pdf_path = "Auswertungen1.pdf"
doc = fitz.open(pdf_path)

for page_num in range(len(doc)):

    page = doc.load_page(page_num)
    pix = page.get_pixmap(dpi=300)  # Increase DPI for better quality
    output_path = f"Ausertungen{page_num + 1}.png"
    pix.save(output_path)

doc.close()
```

```python
import ollama

response = ollama.chat(
    model='llama3.2-vision',
    messages=[{
        'role': 'user',
        'content': 'You are given 4 pages of pdf as input. Recap and write all the information you can read. Oragnize your answer as a report at the end.',
        'images': ['Ausertungen1.png', 'Ausertungen2.png', 'Ausertungen3.png', 'Ausertungen4.png']
    }]
)

print(response)
```
